# Task D: Semimajor-Axis Lyapunov-Proxy Prediction

This notebook runs the first Task D pipeline stage:

- build `tau_proxy = min_ij min{t: R_ij(t) >= 10000}` labels
- verify the actual grouping / valid-pair structure in `dataset_v3/sampling_2`
- build initial-condition tabular features
- fit a constant-median baseline and a `RandomForestRegressor`
- save plots and metrics under `results/task_d_initial_lyapunov_proxy/`

Important note: for `sampling_2`, the local dataset is organized as **32 trajectories per nominal system**, not 8. The default physically meaningful infinitesimal pairs are therefore `(i, i+16)` for `i = 0..15`.
